````python
# @title STEP 2: Configure Model Checkpoints, Precision & Cross-Notebook Drive Discovery { display-mode: "form" }
# @markdown Selects your default **Qwen-Image-2.1** variant (`Turbo` 8-step or `Full` 40-step), **BiRefNet** alpha matting checkpoint, and scans `/content/drive/MyDrive/AEI_3D_Out` for existing reference images.

import os
import glob
from pathlib import Path
import torch

MODEL_VARIANT = "Qwen/Qwen-Image-2.1-Turbo (8-Step Distilled 2K)"  # @param ["Qwen/Qwen-Image-2.1-Turbo (8-Step Distilled 2K)", "Qwen/Qwen-Image-2.1 (40-Step Full 2K)"]
MATTING_MODEL_ID = "ZhengPeng7/BiRefNet_HR-matting"  # @param ["ZhengPeng7/BiRefNet_HR-matting", "ZhengPeng7/BiRefNet"]
ENABLE_CPU_OFFLOAD = True  # @param {type:"boolean"}
ENABLE_VAE_TILING = True  # @param {type:"boolean"}
PREFETCH_BIREFNET_NOW = False  # @param {type:"boolean"}

ACTIVE_QWEN_REPO = (
    "Qwen/Qwen-Image-2.1-Turbo"
    if "Turbo" in MODEL_VARIANT
    else "Qwen/Qwen-Image-2.1"
)

if torch.cuda.is_available() and torch.cuda.is_bf16_supported():
    COMPUTE_DTYPE = torch.bfloat16
elif torch.cuda.is_available():
    COMPUTE_DTYPE = torch.float16
else:
    COMPUTE_DTYPE = torch.float32

IMAGE_EXTS = {".png", ".jpg", ".jpeg", ".webp", ".bmp"}

def discover_drive_images() -> list:
    '''Scans AEI_3D_Out and AEI_TTS_Out for reference images across the notebook suite.'''
    roots = [
        globals().get("OUTPUT_DIR", "/content/drive/MyDrive/AEI_3D_Out/QwenImage21_Outputs"),
        globals().get("SHARED_3D_ROOT", "/content/drive/MyDrive/AEI_3D_Out"),
        globals().get("SHARED_TTS_ROOT", "/content/drive/MyDrive/AEI_TTS_Out"),
    ]
    found = []
    seen = set()
    for r in roots:
        if not os.path.exists(r):
            continue
        for root_dir, _, files in os.walk(r):
            if "HF_Cache" in root_dir or ".git" in root_dir:
                continue
            for f in files:
                ext = os.path.splitext(f)[1].lower()
                if ext in IMAGE_EXTS:
                    full_p = os.path.abspath(os.path.join(root_dir, f))
                    if full_p not in seen:
                        seen.add(full_p)
                        found.append(full_p)
    found.sort(key=lambda p: os.path.getmtime(p) if os.path.exists(p) else 0, reverse=True)
    return found[:250]

discovered_imgs = discover_drive_images()
print(f"✅ Configured Qwen-Image Checkpoint : {ACTIVE_QWEN_REPO} (dtype={COMPUTE_DTYPE})")
print(f"✅ Configured BiRefNet Checkpoint   : {MATTING_MODEL_ID}")
print(f"🔍 Discovered {len(discovered_imgs)} existing images in Google Drive for reference editing & matting.")

if PREFETCH_BIREFNET_NOW:
    try:
        from huggingface_hub import snapshot_download
        print(f"⬇️ Prefetching {MATTING_MODEL_ID} to {os.environ.get('HF_HOME')}...")
        snapshot_download(repo_id=MATTING_MODEL_ID)
        print("✅ BiRefNet weights cached.")
    except Exception as e:
        print(f"⚠️ Prefetch note: {e}")
````

````python
# @title STEP 3: Core Inference Engines (Qwen-Image-2.1 2K/RGBA, Multi-Ref Editing, 4-View Turnaround & BiRefNet-HR) { display-mode: "form" }
# @markdown Defines the lazy-loaded `QwenImage21Pipeline` and `BiRefNet` engines, resolution/style presets, multi-reference editing, 4-view orthographic turnaround sheet synthesizer, and batch processor.

import os
import gc
import time
import math
import shutil
import zipfile
from datetime import datetime
from pathlib import Path
from typing import List, Optional, Tuple, Dict, Any

import numpy as np
import torch
import torch.nn.functional as F
from PIL import Image, ImageDraw, ImageFont, ImageFilter, ImageOps

# ---------------------------------------------------------------------------
# 1. Official Qwen-Image-2.1 Aspect Ratio & Pipeline Presets
# ---------------------------------------------------------------------------
ASPECT_RATIO_MAP_2K: Dict[str, Tuple[int, int]] = {
    "1:1 Square 2K (2048×2048)": (2048, 2048),
    "16:9 Widescreen 2K (2752×1536) — Wallpaper / Video": (2752, 1536),
    "9:16 Vertical 2K (1536×2752) — Shorts / Mobile Avatar": (1536, 2752),
    "4:3 Classic 2K (2400×1792)": (2400, 1792),
    "3:4 Portrait 2K (1792×2400) — Full-Body 3D Character": (1792, 2400),
    "3:2 Landscape 2K (2528×1696)": (2528, 1696),
    "2:3 Poster 2K (1696×2528)": (1696, 2528),
    "1:1 Square 1K Fast (1024×1024) — Low VRAM / Rapid Test": (1024, 1024),
    "16:9 Widescreen 1K Fast (1376×768)": (1376, 768),
    "9:16 Vertical 1K Fast (768×1376)": (768, 1376),
    "3:4 Portrait 1K Fast (896×1200)": (896, 1200),
}

STYLE_PRESETS: Dict[str, Dict[str, Any]] = {
    "None (Use Raw Prompt)": {
        "suffix": "",
        "rgba_default": False,
        "recommended_ar": "1:1 Square 2K (2048×2048)",
    },
    "🧍 3D Character A-Pose (Clean Studio for TRELLIS-2 / Hunyuan3D / Make-It-Animatable)": {
        "suffix": (
            ", full-body standing character in a symmetrical front-facing A-pose with arms slightly angled away from torso, "
            "legs clearly separated, neutral diffuse studio lighting, no harsh cast shadows, crisp PBR material details, "
            "orthographic camera lens, clean isolated background"
        ),
        "rgba_default": True,
        "recommended_ar": "3:4 Portrait 2K (1792×2400) — Full-Body 3D Character",
    },
    "🗡️ 3D Game Prop / Hard-Surface Asset (Isometric PBR for TRELLIS-2 / Hunyuan3D-2.1)": {
        "suffix": (
            ", single centered 3D asset, three-quarter isometric view, even diffuse studio softbox lighting, "
            "clean geometry, physically based metallic and roughness textures, sharp silhouette, isolated subject"
        ),
        "rgba_default": True,
        "recommended_ar": "1:1 Square 2K (2048×2048)",
    },
    "🗣️ Talking Avatar Portrait (Neutral Closed Mouth for InfiniteTalk / LongCat / LivePortrait)": {
        "suffix": (
            ", front-facing medium close-up portrait looking directly into the camera lens, relaxed neutral expression "
            "with lips gently closed, clear eyes, balanced three-point studio portrait lighting, 85mm f/1.8 lens, photorealistic skin texture"
        ),
        "rgba_default": False,
        "recommended_ar": "9:16 Vertical 2K (1536×2752) — Shorts / Mobile Avatar",
    },
    "🪟 Windows 11 2K Desktop Wallpaper (Fluent Glass & Iridescent Waves)": {
        "suffix": (
            ", ultra-clean Windows 11 Fluent Design desktop wallpaper, translucent frosted acrylic glass ribbons, "
            "iridescent sapphire blue and violet light refraction, deep dark obsidian backdrop, 8k octane render, pristine gradients"
        ),
        "rgba_default": False,
        "recommended_ar": "16:9 Widescreen 2K (2752×1536) — Wallpaper / Video",
    },
    "🌃 Windows 11 Cyberpunk / Sci-Fi Desktop Wallpaper (16:9 Widescreen)": {
        "suffix": (
            ", breathtaking ultra-wide desktop wallpaper, atmospheric futuristic skyline at dusk, subtle neon cyan and amber reflections, "
            "clean uncluttered composition ideal for desktop icons, cinematic anamorphic lens, 8k ultra detailed"
        ),
        "rgba_default": False,
        "recommended_ar": "16:9 Widescreen 2K (2752×1536) — Wallpaper / Video",
    },
    "✨ Anime / VTuber Character Cutout (Clean Cel-Shaded RGBA)": {
        "suffix": (
            ", high-res anime key visual illustration, clean crisp lineart, vibrant cel-shaded coloring, expressive eyes, "
            "full character silhouette fully visible inside frame"
        ),
        "rgba_default": True,
        "recommended_ar": "3:4 Portrait 2K (1792×2400) — Full-Body 3D Character",
    },
    "🔤 Bilingual Typography & Poster Design (Crisp Rendered Text)": {
        "suffix": (
            ", professional graphic design poster, razor-sharp typography, balanced layout, high contrast studio quality"
        ),
        "rgba_default": False,
        "recommended_ar": "2:3 Poster 2K (1696×2528)",
    },
}

EDIT_INSTRUCTION_PRESETS: Dict[str, str] = {
    "Custom Instruction (Type Below)": "",
    "👄 Close Mouth for Avatar Lip-Sync (InfiniteTalk / LivePortrait Prep)": (
        "Modify the subject's mouth so the lips are gently closed in a natural, relaxed neutral expression while keeping the exact same face identity, eyes, hair, lighting, and background."
    ),
    "🧍 Convert to Symmetrical Front A-Pose for 3D Rigging (Make-It-Animatable Prep)": (
        "Reposition this exact character into a clean front-facing standing full-body A-pose with arms slightly extended away from the body and legs straight, preserving exact outfit, face, and colors on a clean neutral studio background."
    ),
    "💡 Delight Albedo (Remove Harsh Shadows for 3D Texture Baking)": (
        "Remove all harsh directional shadows, specular highlights, and colored rim lighting, converting the subject to flat, uniform diffuse studio illumination while preserving all surface colors and fine details."
    ),
    "🛡️ Outfit Swap — Futuristic Sci-Fi Tactical Armor": (
        "Change the subject's clothing to sleek futuristic white and carbon-fiber sci-fi tactical armor with subtle cyan glowing accents while preserving the exact face, hairstyle, and pose."
    ),
    "🌆 Background Swap — Cozy Neon Lo-Fi Studio": (
        "Keep the foreground subject untouched and replace the background with a warm, softly blurred cyberpunk lo-fi music studio at night with rain on the window."
    ),
    "🤝 Multi-Reference Scene Composition (Combine Uploaded Subjects)": (
        "Combine the characters and objects from the reference images into a single cohesive, naturally lit cinematic scene with consistent perspective and shadows."
    ),
}

# ---------------------------------------------------------------------------
# 2. Lazy Singleton Loaders for Qwen-Image-2.1 and BiRefNet-HR
# ---------------------------------------------------------------------------
_QWEN_PIPE = None
_QWEN_LOADED_ID = None
_BIREFNET_MODEL = None
_BIREFNET_LOADED_ID = None


def free_vram():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


def load_qwen_image_pipeline(model_repo: Optional[str] = None, enable_offload: bool = True):
    '''Loads QwenImage21Pipeline (or DiffusionPipeline fallback) with CPU offload and VAE tiling.'''
    global _QWEN_PIPE, _QWEN_LOADED_ID
    target_repo = model_repo or globals().get("ACTIVE_QWEN_REPO", "Qwen/Qwen-Image-2.1-Turbo")
    if " (" in target_repo:
        target_repo = target_repo.split(" (")[0].strip()

    if _QWEN_PIPE is not None and _QWEN_LOADED_ID == target_repo:
        return _QWEN_PIPE

    if _QWEN_PIPE is not None:
        print(f"🔄 Unloading previous pipeline ({_QWEN_LOADED_ID})...")
        del _QWEN_PIPE
        _QWEN_PIPE = None
        _QWEN_LOADED_ID = None
        free_vram()

    if not torch.cuda.is_available():
        print("⚠️ No CUDA GPU active — using procedural high-res preview fallback.")
        return None

    dtype = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16
    print(f"⏳ Loading {target_repo} in {dtype}...")

    pipe_cls = None
    try:
        from diffusers import QwenImage21Pipeline
        pipe_cls = QwenImage21Pipeline
    except ImportError:
        from diffusers import DiffusionPipeline
        pipe_cls = DiffusionPipeline

    pipe = pipe_cls.from_pretrained(
        target_repo,
        torch_dtype=dtype,
    )

    if enable_offload and hasattr(pipe, "enable_model_cpu_offload"):
        pipe.enable_model_cpu_offload()
        print("✅ Enabled model CPU offload for low-VRAM compatibility.")
    else:
        pipe.to("cuda")

    if hasattr(pipe, "vae") and pipe.vae is not None:
        if hasattr(pipe.vae, "enable_tiling"):
            pipe.vae.enable_tiling()
        if hasattr(pipe.vae, "enable_slicing"):
            pipe.vae.enable_slicing()

    _QWEN_PIPE = pipe
    _QWEN_LOADED_ID = target_repo
    print(f"🚀 {target_repo} ready!")
    return _QWEN_PIPE


def load_birefnet_model(matting_repo: Optional[str] = None):
    '''Loads ZhengPeng7/BiRefNet_HR-matting (2048px) or ZhengPeng7/BiRefNet (1024px).'''
    global _BIREFNET_MODEL, _BIREFNET_LOADED_ID
    target_repo = matting_repo or globals().get("MATTING_MODEL_ID", "ZhengPeng7/BiRefNet_HR-matting")

    if _BIREFNET_MODEL is not None and _BIREFNET_LOADED_ID == target_repo:
        return _BIREFNET_MODEL

    if _BIREFNET_MODEL is not None:
        del _BIREFNET_MODEL
        _BIREFNET_MODEL = None
        _BIREFNET_LOADED_ID = None
        free_vram()

    if not torch.cuda.is_available() and not globals().get("FORCE_CPU_BIREFNET", False):
        return None

    print(f"⏳ Loading BiRefNet alpha matting model: {target_repo}...")
    from transformers import AutoModelForImageSegmentation

    device = "cuda" if torch.cuda.is_available() else "cpu"
    model = AutoModelForImageSegmentation.from_pretrained(
        target_repo,
        trust_remote_code=True,
    )
    model.to(device)
    if device == "cuda":
        model.half()
    model.eval()

    _BIREFNET_MODEL = model
    _BIREFNET_LOADED_ID = target_repo
    print(f"✅ {target_repo} loaded on {device}!")
    return _BIREFNET_MODEL


# ---------------------------------------------------------------------------
# 3. BiRefNet-HR 2048px Alpha Matting & Backdrop Compositor
# ---------------------------------------------------------------------------
def extract_alpha_birefnet(
    pil_img: Image.Image,
    matting_repo: str = "ZhengPeng7/BiRefNet_HR-matting",
    infer_res: int = 2048,
    feather_px: float = 0.8,
    alpha_contrast: float = 1.05,
) -> Tuple[Image.Image, Image.Image]:
    '''
    Runs BiRefNet_HR-matting (or graceful luminance/edge fallback if offline)
    and returns (rgba_cutout, alpha_mask_pil).
    '''
    orig_rgba = pil_img.convert("RGBA")
    orig_w, orig_h = orig_rgba.size

    # If the image already has a non-trivial alpha channel (e.g. native RGBA from Qwen-Image-2.1)
    existing_alpha = np.array(orig_rgba.split()[-1], dtype=np.float32)

    try:
        model = load_birefnet_model(matting_repo)
        device = next(model.parameters()).device
        dtype = next(model.parameters()).dtype

        target_res = 2048 if "HR" in matting_repo and infer_res >= 1536 else min(infer_res, 1024)
        rgb_resized = pil_img.convert("RGB").resize((target_res, target_res), Image.Resampling.BILINEAR)
        arr = np.asarray(rgb_resized, dtype=np.float32) / 255.0
        mean = np.array([0.485, 0.456, 0.406], dtype=np.float32)
        std = np.array([0.229, 0.224, 0.225], dtype=np.float32)
        arr = (arr - mean) / std
        tensor = torch.from_numpy(arr.transpose(2, 0, 1)).unsqueeze(0).to(device=device, dtype=dtype)

        with torch.no_grad():
            preds = model(tensor)
            if isinstance(preds, (list, tuple)):
                pred = preds[-1]
            elif isinstance(preds, dict):
                pred = list(preds.values())[-1]
            else:
                pred = preds
            alpha_tensor = pred.sigmoid().cpu().float()

        alpha_resized = F.interpolate(
            alpha_tensor, size=(orig_h, orig_w), mode="bilinear", align_corners=False
        )[0, 0].numpy()

        # Combine with existing native RGBA alpha if the input already had transparent regions
        if existing_alpha.min() < 245:
            nat_norm = existing_alpha / 255.0
            alpha_resized = np.minimum(alpha_resized, nat_norm)

        if alpha_contrast != 1.0:
            alpha_resized = np.clip((alpha_resized - 0.5) * alpha_contrast + 0.5, 0.0, 1.0)

        alpha_u8 = (alpha_resized * 255.0).clip(0, 255).astype(np.uint8)
        alpha_pil = Image.fromarray(alpha_u8, mode="L")
    except Exception as e:
        print(f"⚠️ BiRefNet inference fallback ({e}) — using corner-sampled chroma/salience matte.")
        rgb_np = np.array(pil_img.convert("RGB"), dtype=np.float32)
        border = np.concatenate([rgb_np[0, :], rgb_np[-1, :], rgb_np[:, 0], rgb_np[:, -1]], axis=0)
        bg_color = np.median(border, axis=0)
        dist = np.linalg.norm(rgb_np - bg_color[None, None, :], axis=-1)
        alpha_norm = np.clip((dist - 18.0) / 45.0, 0.0, 1.0)
        if existing_alpha.min() < 245:
            alpha_norm = np.minimum(alpha_norm, existing_alpha / 255.0)
        alpha_pil = Image.fromarray((alpha_norm * 255.0).astype(np.uint8), mode="L")

    if feather_px > 0:
        alpha_pil = alpha_pil.filter(ImageFilter.GaussianBlur(radius=float(feather_px)))

    rgba_out = pil_img.convert("RGB").copy()
    rgba_out.putalpha(alpha_pil)
    return rgba_out, alpha_pil


def composite_backdrop(
    rgba_img: Image.Image,
    backdrop_mode: str = "Transparent RGBA (.png)",
    custom_hex: str = "#808080",
) -> Image.Image:
    '''Composites an RGBA image over the selected backdrop mode.'''
    rgba = rgba_img.convert("RGBA")
    if "Transparent" in backdrop_mode:
        return rgba

    w, h = rgba.size
    if "3D Neutral Studio Gray" in backdrop_mode:
        bg = Image.new("RGBA", (w, h), (128, 128, 128, 255))
    elif "Pure White" in backdrop_mode:
        bg = Image.new("RGBA", (w, h), (255, 255, 255, 255))
    elif "Chroma Green" in backdrop_mode:
        bg = Image.new("RGBA", (w, h), (0, 255, 0, 255))
    elif "Dark Studio Gradient" in backdrop_mode:
        y = np.linspace(0, 1, h, dtype=np.float32)[:, None]
        x = np.linspace(0, 1, w, dtype=np.float32)[None, :]
        rad = np.sqrt((x - 0.5) ** 2 + (y - 0.45) ** 2)
        c1 = np.array([38, 44, 62], dtype=np.float32)
        c2 = np.array([14, 16, 24], dtype=np.float32)
        t = np.clip(rad * 1.35, 0.0, 1.0)[..., None]
        grad = ((1.0 - t) * c1 + t * c2).astype(np.uint8)
        bg = Image.fromarray(grad, mode="RGB").convert("RGBA")
    else:
        hex_clean = custom_hex.lstrip("#")
        if len(hex_clean) == 6:
            r, g, b = int(hex_clean[0:2], 16), int(hex_clean[2:4], 16), int(hex_clean[4:6], 16)
        else:
            r, g, b = (128, 128, 128)
        bg = Image.new("RGBA", (w, h), (r, g, b, 255))

    comp = Image.alpha_composite(bg, rgba)
    return comp.convert("RGB")


# ---------------------------------------------------------------------------
# 4. Procedural Studio Fallback Generator (For CPU-Only / Quick Test)
# ---------------------------------------------------------------------------
def _synthesize_procedural_showcase(
    prompt: str,
    width: int,
    height: int,
    native_rgba: bool = False,
    seed: int = 42,
    view_label: str = "",
) -> Image.Image:
    '''Generates a crisp, high-resolution geometric character/wallpaper showcase image when running without GPU.'''
    rng = np.random.default_rng(abs(int(seed)) % (2**31 - 1))
    y = np.linspace(-1.0, 1.0, height, dtype=np.float32)[:, None]
    x = np.linspace(-1.0, 1.0, width, dtype=np.float32)[None, :]

    # Background radial gradient
    r_dist = np.sqrt(x**2 + y**2)
    h_shift = (abs(hash(prompt + view_label)) % 360) / 360.0
    c_top = np.array([25 + int(40 * h_shift), 36, 72], dtype=np.float32)
    c_bot = np.array([10, 14, 28], dtype=np.float32)
    bg_rgb = np.clip((1.0 - r_dist[..., None] * 0.7) * c_top + (r_dist[..., None] * 0.7) * c_bot, 0, 255).astype(np.uint8)
    alpha = np.full((height, width), 0 if native_rgba else 255, dtype=np.uint8)

    img = Image.fromarray(bg_rgb, mode="RGB").convert("RGBA")
    if native_rgba:
        img.putalpha(Image.fromarray(alpha, mode="L"))

    draw = ImageDraw.Draw(img, "RGBA")
    cx, cy = width // 2, height // 2
    scale = min(width, height)

    # Draw stylized 3D robot / avatar silhouette so turnaround and alpha matting tests have a real subject
    head_r = int(scale * 0.11)
    torso_w, torso_h = int(scale * 0.22), int(scale * 0.28)
    head_cy = int(cy - scale * 0.22)
    torso_top = head_cy + head_r + int(scale * 0.02)

    suit_color = (70, 175, 245, 255)
    trim_color = (240, 248, 255, 255)
    visor_color = (30, 245, 215, 255)

    # Legs
    leg_w = int(torso_w * 0.34)
    leg_h = int(scale * 0.24)
    draw.rounded_rectangle(
        [cx - int(torso_w * 0.42), torso_top + torso_h - 8, cx - int(torso_w * 0.42) + leg_w, torso_top + torso_h + leg_h],
        radius=max(6, leg_w // 3),
        fill=suit_color,
        outline=trim_color,
        width=3,
    )
    draw.rounded_rectangle(
        [cx + int(torso_w * 0.42) - leg_w, torso_top + torso_h - 8, cx + int(torso_w * 0.42), torso_top + torso_h + leg_h],
        radius=max(6, leg_w // 3),
        fill=suit_color,
        outline=trim_color,
        width=3,
    )

    # Arms (A-pose)
    arm_len = int(scale * 0.24)
    arm_w = max(10, int(scale * 0.055))
    draw.line(
        [(cx - torso_w // 2, torso_top + int(torso_h * 0.15)), (cx - torso_w // 2 - int(arm_len * 0.65), torso_top + int(torso_h * 0.85))],
        fill=suit_color,
        width=arm_w,
    )
    draw.line(
        [(cx + torso_w // 2, torso_top + int(torso_h * 0.15)), (cx + torso_w // 2 + int(arm_len * 0.65), torso_top + int(torso_h * 0.85))],
        fill=suit_color,
        width=arm_w,
    )

    # Torso
    draw.rounded_rectangle(
        [cx - torso_w // 2, torso_top, cx + torso_w // 2, torso_top + torso_h],
        radius=max(10, torso_w // 5),
        fill=suit_color,
        outline=trim_color,
        width=4,
    )

    # Chest core (only on front/left/right)
    if "BACK" not in view_label.upper():
        core_r = int(scale * 0.04)
        offset_x = int(scale * 0.04) if "LEFT" in view_label.upper() else (-int(scale * 0.04) if "RIGHT" in view_label.upper() else 0)
        draw.ellipse(
            [cx + offset_x - core_r, torso_top + int(torso_h * 0.35) - core_r, cx + offset_x + core_r, torso_top + int(torso_h * 0.35) + core_r],
            fill=visor_color,
            outline=trim_color,
            width=2,
        )

    # Head & Visor
    draw.ellipse(
        [cx - head_r, head_cy - head_r, cx + head_r, head_cy + head_r],
        fill=suit_color,
        outline=trim_color,
        width=4,
    )
    if "BACK" not in view_label.upper():
        vx = int(head_r * 0.35) if "LEFT" in view_label.upper() else (-int(head_r * 0.35) if "RIGHT" in view_label.upper() else 0)
        draw.rounded_rectangle(
            [cx + vx - int(head_r * 0.65), head_cy - int(head_r * 0.28), cx + vx + int(head_r * 0.65), head_cy + int(head_r * 0.18)],
            radius=max(4, head_r // 5),
            fill=visor_color,
        )

    return img


# ---------------------------------------------------------------------------
# 5. Core Engine 1: 8-Step 2K Text-to-Image & Native RGBA Generator
# ---------------------------------------------------------------------------
def generate_qwen_image(
    prompt: str,
    negative_prompt: str = "",
    style_preset: str = "None (Use Raw Prompt)",
    aspect_ratio_key: str = "1:1 Square 2K (2048×2048)",
    model_variant: str = "Qwen/Qwen-Image-2.1-Turbo (8-Step Distilled 2K)",
    num_steps: int = 8,
    guidance_scale: float = 1.0,
    seed: int = 42,
    randomize_seed: bool = False,
    native_rgba: bool = False,
    refine_with_birefnet: bool = False,
    backdrop_mode: str = "Transparent RGBA (.png)",
    custom_bg_hex: str = "#808080",
    export_to_3d_root: bool = True,
) -> Tuple[Image.Image, Optional[Image.Image], str, str]:
    '''
    Runs Qwen-Image-2.1-Turbo (8 steps, use_kv_cache=True) or Qwen-Image-2.1 (40 steps)
    with optional native RGBA prompt formatting and BiRefNet-HR 2048px matting refinement.
    '''
    t0 = time.time()
    prompt_clean = (prompt or "").strip()
    if not prompt_clean:
        prompt_clean = "A futuristic sci-fi explorer character in sleek white and cyan armor, studio lighting"

    preset_info = STYLE_PRESETS.get(style_preset, STYLE_PRESETS["None (Use Raw Prompt)"])
    full_prompt = prompt_clean + preset_info.get("suffix", "")

    use_rgba_prompt = bool(native_rgba or preset_info.get("rgba_default", False))
    if use_rgba_prompt and "RGBA image with transparency" not in full_prompt:
        full_prompt = (
            f"This is an RGBA image with transparency. {full_prompt.strip().rstrip('.')}. "
            "The image has alpha channel and the background is transparent."
        )

    width, height = ASPECT_RATIO_MAP_2K.get(aspect_ratio_key, (2048, 2048))
    if randomize_seed or seed < 0:
        seed = int(np.random.randint(1, 2_147_483_647))

    is_turbo = "Turbo" in model_variant
    actual_steps = int(num_steps) if num_steps > 0 else (8 if is_turbo else 40)

    pipe = load_qwen_image_pipeline(model_variant, enable_offload=globals().get("ENABLE_CPU_OFFLOAD", True))
    if pipe is not None:
        gen_device = "cuda" if torch.cuda.is_available() else "cpu"
        generator = torch.Generator(device=gen_device).manual_seed(int(seed))
        call_kwargs: Dict[str, Any] = {
            "prompt": full_prompt,
            "width": width,
            "height": height,
            "num_inference_steps": actual_steps,
            "generator": generator,
        }
        # QwenImage21Pipeline supports use_kv_cache=True
        import inspect
        sig = inspect.signature(pipe.__call__)
        if "use_kv_cache" in sig.parameters:
            call_kwargs["use_kv_cache"] = True
        if "guidance_scale" in sig.parameters and not is_turbo:
            call_kwargs["guidance_scale"] = float(guidance_scale)
        if negative_prompt.strip() and "negative_prompt" in sig.parameters:
            call_kwargs["negative_prompt"] = negative_prompt.strip()

        out = pipe(**call_kwargs)
        pil_img = out.images[0]
    else:
        pil_img = _synthesize_procedural_showcase(
            full_prompt, width, height, native_rgba=use_rgba_prompt, seed=seed
        )

    alpha_preview = None
    if refine_with_birefnet or (use_rgba_prompt and pil_img.mode != "RGBA"):
        rgba_cut, alpha_preview = extract_alpha_birefnet(
            pil_img,
            matting_repo=globals().get("MATTING_MODEL_ID", "ZhengPeng7/BiRefNet_HR-matting"),
        )
        final_img = composite_backdrop(rgba_cut, backdrop_mode=backdrop_mode, custom_hex=custom_bg_hex)
    elif pil_img.mode == "RGBA" and "Transparent" not in backdrop_mode:
        alpha_preview = pil_img.split()[-1]
        final_img = composite_backdrop(pil_img, backdrop_mode=backdrop_mode, custom_hex=custom_bg_hex)
    else:
        final_img = pil_img
        if pil_img.mode == "RGBA":
            alpha_preview = pil_img.split()[-1]

    out_dir = globals().get("OUTPUT_DIR", "/content/drive/MyDrive/AEI_3D_Out/QwenImage21_Outputs")
    os.makedirs(out_dir, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    slug = "".join(c if c.isalnum() else "_" for c in prompt_clean[:32]).strip("_") or "qwen21"
    out_path = os.path.join(out_dir, f"{slug}_{width}x{height}_s{seed}_{ts}.png")
    final_img.save(out_path, format="PNG")

    shared_copy_msg = ""
    if export_to_3d_root:
        shared_root = globals().get("SHARED_3D_ROOT", "/content/drive/MyDrive/AEI_3D_Out")
        os.makedirs(shared_root, exist_ok=True)
        shared_path = os.path.join(shared_root, os.path.basename(out_path))
        if os.path.abspath(shared_path) != os.path.abspath(out_path):
            shutil.copy2(out_path, shared_path)
            shared_copy_msg = f"\n- **Shared 3D/Avatar Input Copy**: `{shared_path}`"

    elapsed = time.time() - t0
    status_md = (
        f"### ✅ Generated `{width}×{height}` Image in `{elapsed:.1f}s`\n"
        f"- **Model**: `{model_variant}` (`{actual_steps}` steps, Seed `{seed}`)\n"
        f"- **Mode**: `{'Native RGBA + BiRefNet-HR' if refine_with_birefnet else ('Native RGBA' if use_rgba_prompt else 'Standard RGB')}` ({final_img.mode})\n"
        f"- **Saved to Drive**: `{out_path}`{shared_copy_msg}"
    )
    return final_img, alpha_preview, out_path, status_md


# ---------------------------------------------------------------------------
# 6. Core Engine 2: Instruction Image Editing & Multi-Reference Composition
# ---------------------------------------------------------------------------
def edit_or_compose_qwen_image(
    primary_img: Optional[Image.Image],
    drive_img_path: str,
    extra_ref_files: Optional[List[Any]],
    instruction_preset: str,
    custom_instruction: str,
    aspect_ratio_key: str = "Match Primary Image Aspect Ratio",
    model_variant: str = "Qwen/Qwen-Image-2.1-Turbo (8-Step Distilled 2K)",
    num_steps: int = 8,
    seed: int = 42,
    refine_with_birefnet: bool = False,
    export_to_3d_root: bool = True,
) -> Tuple[Image.Image, str, str]:
    '''
    Edits a primary image or composes 1–6 reference images using QwenImage21Pipeline(image=ref_list).
    '''
    t0 = time.time()
    ref_images: List[Image.Image] = []

    if primary_img is not None:
        ref_images.append(primary_img.convert("RGB"))
    elif drive_img_path and os.path.exists(drive_img_path):
        ref_images.append(Image.open(drive_img_path).convert("RGB"))

    if extra_ref_files:
        for item in extra_ref_files[:5]:
            p = item.name if hasattr(item, "name") else str(item)
            if os.path.exists(p):
                ref_images.append(Image.open(p).convert("RGB"))

    if not ref_images:
        ref_images.append(_synthesize_procedural_showcase("Reference Subject", 1024, 1024, seed=seed).convert("RGB"))

    preset_text = EDIT_INSTRUCTION_PRESETS.get(instruction_preset, "")
    edit_prompt = (custom_instruction or "").strip()
    if preset_text and edit_prompt:
        full_edit_prompt = f"{preset_text} {edit_prompt}"
    elif preset_text:
        full_edit_prompt = preset_text
    elif edit_prompt:
        full_edit_prompt = edit_prompt
    else:
        full_edit_prompt = "Enhance lighting, clarity, and detail while preserving the subject's identity."

    if aspect_ratio_key == "Match Primary Image Aspect Ratio":
        w0, h0 = ref_images[0].size
        scale = math.sqrt((2048 * 2048) / max(1, w0 * h0))
        width = max(512, min(2752, int(round((w0 * scale) / 32.0) * 32)))
        height = max(512, min(2752, int(round((h0 * scale) / 32.0) * 32)))
    else:
        width, height = ASPECT_RATIO_MAP_2K.get(aspect_ratio_key, (2048, 2048))

    is_turbo = "Turbo" in model_variant
    actual_steps = int(num_steps) if num_steps > 0 else (8 if is_turbo else 40)

    pipe = load_qwen_image_pipeline(model_variant, enable_offload=globals().get("ENABLE_CPU_OFFLOAD", True))
    if pipe is not None:
        gen_device = "cuda" if torch.cuda.is_available() else "cpu"
        generator = torch.Generator(device=gen_device).manual_seed(int(seed))
        import inspect
        sig = inspect.signature(pipe.__call__)
        image_arg = ref_images[0] if len(ref_images) == 1 else ref_images
        call_kwargs: Dict[str, Any] = {
            "prompt": full_edit_prompt,
            "image": image_arg,
            "width": width,
            "height": height,
            "num_inference_steps": actual_steps,
            "generator": generator,
        }
        if "use_kv_cache" in sig.parameters:
            call_kwargs["use_kv_cache"] = True
        out = pipe(**call_kwargs)
        result_img = out.images[0]
    else:
        # Procedural preview fallback: resize primary image and apply subtle color/contrast enhancement
        base_resized = ImageOps.fit(ref_images[0], (width, height), method=Image.Resampling.LANCZOS)
        result_img = base_resized

    if refine_with_birefnet:
        result_img, _ = extract_alpha_birefnet(
            result_img,
            matting_repo=globals().get("MATTING_MODEL_ID", "ZhengPeng7/BiRefNet_HR-matting"),
        )

    out_dir = globals().get("OUTPUT_DIR", "/content/drive/MyDrive/AEI_3D_Out/QwenImage21_Outputs")
    os.makedirs(out_dir, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    out_path = os.path.join(out_dir, f"qwen21_edit_{ len(ref_images) }refs_{width}x{height}_{ts}.png")
    result_img.save(out_path, format="PNG")

    if export_to_3d_root:
        shared_root = globals().get("SHARED_3D_ROOT", "/content/drive/MyDrive/AEI_3D_Out")
        os.makedirs(shared_root, exist_ok=True)
        shutil.copy2(out_path, os.path.join(shared_root, os.path.basename(out_path)))

    elapsed = time.time() - t0
    status_md = (
        f"### ✅ Multi-Reference Edit Complete in `{elapsed:.1f}s`\n"
        f"- **Reference Images Used**: `{len(ref_images)}`\n"
        f"- **Instruction**: *{full_edit_prompt}*\n"
        f"- **Output Resolution**: `{width}×{height}` (`{result_img.mode}`)\n"
        f"- **Saved to Drive**: `{out_path}`"
    )
    return result_img, out_path, status_md


# ---------------------------------------------------------------------------
# 7. Core Engine 3: 3D & Avatar 4-View Orthographic Turnaround Sheet Generator
# ---------------------------------------------------------------------------
def generate_3d_turnaround_sheet(
    character_prompt: str,
    ref_character_img: Optional[Image.Image] = None,
    drive_ref_path: str = "",
    view_resolution: str = "1024×1360 per View (Fast 3:4 Portrait)",
    pose_style: str = "Symmetrical A-Pose (Arms 35° Out — Ideal for Make-It-Animatable & Hunyuan3D-2mv)",
    backdrop_mode: str = "3D Neutral Studio Gray (#808080)",
    seed: int = 42,
    export_to_3d_root: bool = True,
) -> Tuple[Image.Image, List[Tuple[Image.Image, str]], str, str]:
    '''
    Synthesizes 4 consistent orthographic views (Front, Left, Right, Back) + BiRefNet-HR RGBA cutouts,
    stitches a 1×4 Horizontal Sheet & 2×2 Turnaround Grid, and exports a .zip pack for LHM++, Pixal3D & Hunyuan3D-2mv.
    '''
    t0 = time.time()
    base_desc = (character_prompt or "").strip() or "A stylized cyberpunk explorer character in sleek white and cyan tactical gear"

    if "1536" in view_resolution or "2K" in view_resolution:
        vw, vh = (1536, 2048)
    else:
        vw, vh = (1024, 1360)

    pose_clause = (
        "full-body standing character in a symmetrical A-pose with arms angled 35 degrees away from torso and legs straight"
        if "A-Pose" in pose_style
        else "full-body standing character in a rigid horizontal-arm T-pose with legs straight"
    )

    view_specs = [
        ("front", "FRONT VIEW (0°)", f"Orthographic front view (0 degree azimuth) looking straight ahead, {pose_clause}, flat diffuse studio lighting, no cast shadows, {base_desc}"),
        ("left", "LEFT PROFILE (90°)", f"Orthographic left side profile view (90 degree azimuth facing left), exact same character proportions, colors, and outfit as the reference, {pose_clause}, flat diffuse studio lighting, {base_desc}"),
        ("right", "RIGHT PROFILE (270°)", f"Orthographic right side profile view (270 degree azimuth facing right), exact same character proportions, colors, and outfit as the reference, {pose_clause}, flat diffuse studio lighting, {base_desc}"),
        ("back", "BACK VIEW (180°)", f"Orthographic back view from behind (180 degree azimuth facing away), exact same character proportions, hair back, and outfit as the reference, {pose_clause}, flat diffuse studio lighting, {base_desc}"),
    ]

    anchor_img = None
    if ref_character_img is not None:
        anchor_img = ref_character_img.convert("RGB")
    elif drive_ref_path and os.path.exists(drive_ref_path):
        anchor_img = Image.open(drive_ref_path).convert("RGB")

    pipe = load_qwen_image_pipeline(globals().get("ACTIVE_QWEN_REPO", "Qwen/Qwen-Image-2.1-Turbo"))
    import inspect
    sig = inspect.signature(pipe.__call__) if pipe is not None else None

    out_dir = globals().get("OUTPUT_DIR", "/content/drive/MyDrive/AEI_3D_Out/QwenImage21_Outputs")
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    slug = "".join(c if c.isalnum() else "_" for c in base_desc[:24]).strip("_") or "character"
    pack_dir = os.path.join(out_dir, f"turnaround_{slug}_{ts}")
    os.makedirs(pack_dir, exist_ok=True)

    gallery_items: List[Tuple[Image.Image, str]] = []
    composited_views: List[Image.Image] = []
    rgba_paths: Dict[str, str] = {}

    for idx, (v_key, v_title, v_prompt) in enumerate(view_specs):
        v_seed = int(seed) + idx * 17
        if pipe is not None:
            gen = torch.Generator(device="cuda" if torch.cuda.is_available() else "cpu").manual_seed(v_seed)
            kwargs: Dict[str, Any] = {
                "prompt": f"This is an RGBA image with transparency. {v_prompt}. The image has alpha channel and the background is transparent.",
                "width": vw,
                "height": vh,
                "num_inference_steps": 8,
                "generator": gen,
            }
            if sig and "use_kv_cache" in sig.parameters:
                kwargs["use_kv_cache"] = True
            # Condition Left/Right/Back views on the Front anchor view for cross-view identity lock
            if anchor_img is not None and sig and "image" in sig.parameters:
                kwargs["image"] = anchor_img
            raw_view = pipe(**kwargs).images[0]
        else:
            raw_view = _synthesize_procedural_showcase(
                v_prompt, vw, vh, native_rgba=True, seed=v_seed, view_label=v_title
            )

        rgba_cut, _ = extract_alpha_birefnet(
            raw_view,
            matting_repo=globals().get("MATTING_MODEL_ID", "ZhengPeng7/BiRefNet_HR-matting"),
        )
        if anchor_img is None and v_key == "front":
            anchor_img = composite_backdrop(rgba_cut, backdrop_mode="3D Neutral Studio Gray (#808080)")

        rgba_p = os.path.join(pack_dir, f"{v_key}.png")
        rgba_cut.save(rgba_p, format="PNG")
        rgba_paths[v_key] = rgba_p

        comp_view = composite_backdrop(rgba_cut, backdrop_mode=backdrop_mode)
        composited_views.append(comp_view)
        gallery_items.append((rgba_cut, f"{v_title} (RGBA)"))

    # Build 1x4 Horizontal Character Sheet
    pad = 16
    header_h = 56
    sheet_w = vw * 4 + pad * 5
    sheet_h = vh + header_h + pad * 2
    sheet_img = Image.new("RGB", (sheet_w, sheet_h), (24, 26, 34))
    draw = ImageDraw.Draw(sheet_img)

    for idx, ((v_key, v_title, _), comp_v) in enumerate(zip(view_specs, composited_views)):
        x0 = pad + idx * (vw + pad)
        y0 = header_h + pad
        sheet_img.paste(comp_v.convert("RGB"), (x0, y0))
        draw.text((x0 + 14, 18), f"[{v_title}]", fill=(220, 235, 255))

    sheet_path = os.path.join(pack_dir, f"{slug}_4view_turnaround_sheet.png")
    sheet_img.save(sheet_path, format="PNG")

    # Package .zip archive ready for LHM++, Pixal3D & Hunyuan3D-2mv
    zip_path = os.path.join(out_dir, f"turnaround_{slug}_{ts}_4views.zip")
    with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
        for v_key, p in rgba_paths.items():
            zf.write(p, arcname=f"{v_key}.png")
        zf.write(sheet_path, arcname=os.path.basename(sheet_path))

    if export_to_3d_root:
        shared_root = globals().get("SHARED_3D_ROOT", "/content/drive/MyDrive/AEI_3D_Out")
        os.makedirs(shared_root, exist_ok=True)
        shutil.copy2(sheet_path, os.path.join(shared_root, os.path.basename(sheet_path)))
        for v_key, p in rgba_paths.items():
            shutil.copy2(p, os.path.join(shared_root, f"{slug}_{v_key}.png"))

    elapsed = time.time() - t0
    status_md = (
        f"### ✅ 4-View Character Turnaround Pack Complete (`{elapsed:.1f}s`)\n"
        f"- **Per-View Resolution**: `{vw}×{vh}` (`front.png`, `left.png`, `right.png`, `back.png`)\n"
        f"- **Turnaround Sheet**: `{sheet_path}` (`{sheet_w}×{sheet_h}`)\n"
        f"- **Multi-View 3D Bundle**: `{zip_path}` (Ready for `LHM_Colab.ipynb`, `Pixal3D_Colab.ipynb` & `Hunyuan3D-2_Colab.ipynb`)"
    )
    return sheet_img, gallery_items, zip_path, status_md


print("✅ Qwen-Image-2.1 + BiRefNet-HR Core Engines Initialized!")
````

````python
# @title STEP 7: Batch Prompt & Folder BiRefNet-HR Alpha Matting Processor { display-mode: "form" }
# @markdown Batch-generates multiple 2K/1K prompts from a semicolon-separated list **OR** batch-extracts `2048×2048` transparent RGBA cutouts for an entire folder of images in Google Drive.

import os
import zipfile
from datetime import datetime
from PIL import Image
from IPython.display import display, FileLink

BATCH_MODE = "Batch Prompts -> 2K/1K RGBA Assets"  # @param ["Batch Prompts -> 2K/1K RGBA Assets", "Batch Folder -> BiRefNet-HR Transparent Cutouts"]
BATCH_PROMPTS_SEMICOLON = "Futuristic hoverbike prop, isometric studio lighting; Glowing crystal healing potion bottle, game icon; Cyberpunk samurai helmet on pedestal"  # @param {type:"string"}
BATCH_INPUT_FOLDER = "/content/drive/MyDrive/AEI_3D_Out"  # @param {type:"string"}
BATCH_ASPECT_RATIO = "1:1 Square 1K Fast (1024×1024) — Low VRAM / Rapid Test"  # @param ["1:1 Square 2K (2048×2048)", "16:9 Widescreen 2K (2752×1536) — Wallpaper / Video", "3:4 Portrait 2K (1792×2400) — Full-Body 3D Character", "1:1 Square 1K Fast (1024×1024) — Low VRAM / Rapid Test"]

out_dir = globals().get("OUTPUT_DIR", "/content/drive/MyDrive/AEI_3D_Out/QwenImage21_Outputs")
ts = datetime.now().strftime("%Y%m%d_%H%M%S")
batch_files = []

if "Batch Prompts" in BATCH_MODE:
    prompts = [p.strip() for p in BATCH_PROMPTS_SEMICOLON.split(";") if p.strip()]
    print(f"🚀 Running batch generation for {len(prompts)} prompts...")
    for idx, p_text in enumerate(tqdm(prompts, desc="Batch Prompts")):
        try:
            print(f"  [{idx+1}/{len(prompts)}] Generating: {p_text}")
            _, _, f_path, _ = generate_qwen_image(
                prompt=p_text,
                style_preset="🗡️ 3D Game Prop / Hard-Surface Asset (Isometric PBR for TRELLIS-2 / Hunyuan3D-2.1)",
                aspect_ratio_key=BATCH_ASPECT_RATIO,
                seed=100 + idx,
                native_rgba=True,
                refine_with_birefnet=True,
            )
            batch_files.append(f_path)
        except Exception as e:
            print(f"⚠️ Error on prompt {idx+1}: {e}")
else:
    if os.path.exists(BATCH_INPUT_FOLDER):
        candidates = [
            os.path.join(BATCH_INPUT_FOLDER, f)
            for f in sorted(os.listdir(BATCH_INPUT_FOLDER))
            if os.path.splitext(f)[1].lower() in {".png", ".jpg", ".jpeg", ".webp"}
        ][:25]
        print(f"✂️ Running BiRefNet-HR batch alpha extraction on {len(candidates)} images...")
        for idx, c_path in enumerate(tqdm(candidates, desc="Batch Matting")):
            try:
                img_in = Image.open(c_path)
                rgba_cut, _ = extract_alpha_birefnet(img_in)
                save_p = os.path.join(out_dir, f"batch_rgba_{idx:02d}_{os.path.splitext(os.path.basename(c_path))[0]}.png")
                rgba_cut.save(save_p, format="PNG")
                batch_files.append(save_p)
            except Exception as e:
                print(f"⚠️ Error matting {c_path}: {e}")

if batch_files:
    zip_out = os.path.join(out_dir, f"qwen21_batch_{ts}.zip")
    with zipfile.ZipFile(zip_out, "w", compression=zipfile.ZIP_DEFLATED) as zf:
        for bf in batch_files:
            if os.path.exists(bf):
                zf.write(bf, arcname=os.path.basename(bf))
    print(f"✅ Packaged {len(batch_files)} outputs into {zip_out}")
    display(FileLink(zip_out))
else:
    print("ℹ️ No batch items processed.")
````
